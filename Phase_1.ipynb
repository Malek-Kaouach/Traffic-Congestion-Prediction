## Cell 4 — Filter to Bay Area 2017
From the 7.7M-row dataset, keep only:
- Records in the 9 SF Bay Area counties
- Records from calendar year 2017 (matches PEMS-BAY window)
- Only the 18 columns needed for this project

> ⏱️ **Runtime: 1–3 minutes** (chunked CSV read).

```python
import pandas as pd
import glob
import os

# Define local folder path if not already defined
dest_folder = './traffic_project'

# Auto-find the CSV (name varies by Kaggle version)
csv_files = glob.glob(f"{dest_folder}/US_Accidents*.csv")
if not csv_files:
    raise FileNotFoundError(f"No US_Accidents CSV file found in {dest_folder}. Run the download step first.")

INPUT_PATH = csv_files[0]
OUTPUT_PATH = f"{dest_folder}/us_accidents_bay_2017.csv"
print(f"Input file found: {INPUT_PATH}")

KEEP_COLS = [
    'ID', 'Severity', 'Start_Time', 'End_Time',
    'Start_Lat', 'Start_Lng',
    'Distance(mi)', 'City', 'County', 'State',
    'Temperature(F)', 'Humidity(%)', 'Pressure(in)',
    'Visibility(mi)', 'Wind_Speed(mph)', 'Precipitation(in)',
    'Weather_Condition', 'Sunrise_Sunset'
]

BAY_COUNTIES = [
    'Santa Clara', 'Alameda', 'San Mateo',
    'San Francisco', 'Contra Costa',
    'Solano', 'Marin', 'Napa', 'Sonoma'
]

print("\nFiltering...")
chunks = []
total_rows = 0

for i, chunk in enumerate(pd.read_csv(
        INPUT_PATH, usecols=KEEP_COLS,
        chunksize=500_000, low_memory=False)):
    total_rows += len(chunk)
    chunk['Start_Time'] = pd.to_datetime(chunk['Start_Time'], errors='coerce')
    chunk = chunk[(chunk['Start_Time'] >= '2017-01-01') &
                  (chunk['Start_Time'] <  '2018-01-01')]
    chunk = chunk[(chunk['State'] == 'CA') & (chunk['County'].isin(BAY_COUNTIES))]
    if len(chunk) > 0:
        chunks.append(chunk)
    print(f"  Chunk {i+1}: scanned {total_rows:,}, kept {sum(len(c) for c in chunks):,}")

result = pd.concat(chunks, ignore_index=True)
result.to_csv(OUTPUT_PATH, index=False)

print(f"\n✅ Saved locally to: {OUTPUT_PATH}")
print(f"   Rows kept: {len(result):,}")
print(f"   Date range: {result['Start_Time'].min()} → {result['Start_Time'].max()}")
print(f"\n   Counties breakdown:")
print(result['County'].value_counts())

# File size check (Cross-platform Python alternative to `!ls -lh`)
file_size_mb = os.path.getsize(OUTPUT_PATH) / (1024 * 1024)
print(f"\n   File size: {file_size_mb:.2f} MB")
```

## Cell 8 — Load PEMS-BAY Speed Data & Sensor Metadata
Load the speed matrix (52,116 timesteps × 325 sensors) and extract sensor GPS coordinates from the metadata file.

```python
import h5py
import numpy as np
import pandas as pd
import os

# Set local project folder
dest_folder = './traffic_project'

pems_h5 = os.path.join(dest_folder, 'pems-bay.h5')
meta_h5 = os.path.join(dest_folder, 'pems-bay-meta.h5')

# Check if required files exist
if not os.path.exists(pems_h5) or not os.path.exists(meta_h5):
    raise FileNotFoundError(
        f"Missing required HDF5 files in {dest_folder}.\n"
        f"Please ensure both 'pems-bay.h5' and 'pems-bay-meta.h5' are placed in {dest_folder}."
    )

# Load speed data
with h5py.File(pems_h5, 'r') as f:
    speeds = f['speed/block0_values'][:]       # (52116, 325)
    sensor_ids = f['speed/block0_items'][:]    # 325 sensor IDs
    time_ns = f['speed/axis1'][:]              # timestamps in nanoseconds

timestamps = pd.to_datetime(time_ns, unit='ns')
print(f"Speed matrix: {speeds.shape}")
print(f"Time range: {timestamps.min()} → {timestamps.max()}")
print(f"Sensors: {len(sensor_ids)}")

# Load sensor GPS coordinates from meta file
with h5py.File(meta_h5, 'r') as f:
    meta_cols = [c.decode() for c in f['meta/block0_items'][:]]
    meta_vals = f['meta/block0_values'][:]
    meta_sensor_ids = f['meta/axis1'][:]

meta_df = pd.DataFrame(meta_vals, columns=meta_cols)
meta_df['sensor_id'] = meta_sensor_ids
meta_df = meta_df[['sensor_id', 'Latitude', 'Longitude']]
print(f"\nSensor metadata:")
print(meta_df.head())
```

## Cell 9 — Reshape to Long Format
Convert the (52,116 × 325) speed matrix into a long-format DataFrame:
one row per (sensor, timestamp) = ~16.9M rows.
Attach GPS coordinates to each row.

```python
# Convert (52116 × 325) matrix into a long format:
# one row per (sensor, timestamp) = 16.9M rows
print("Reshaping speed matrix into long format...")

# Build the long dataframe
long_df = pd.DataFrame(speeds, index=timestamps, columns=sensor_ids)
long_df.index.name = 'timestamp'
long_df.columns.name = 'sensor_id'
long_df = long_df.stack().reset_index()
long_df.columns = ['timestamp', 'sensor_id', 'speed']

print(f"Long format shape: {long_df.shape}")
print(long_df.head())

# Attach GPS coordinates
long_df = long_df.merge(meta_df, on='sensor_id', how='left')
print(f"\nWith GPS attached: {long_df.shape}")
print(f"Rows missing GPS: {long_df['Latitude'].isna().sum()}\n")
print(long_df.head())
```

## Cell 10 — Load & Prepare Accident Data
Load the filtered Bay Area accident records and extract weather observations.
Each accident record carries certified weather readings from official meteorological stations.

```python
# Set local project directory
dest_folder = './traffic_project'
acc_file = os.path.join(dest_folder, 'us_accidents_bay_2017.csv')

if not os.path.exists(acc_file):
    raise FileNotFoundError(f"Could not find {acc_file}. Please run Cell 4 first to generate the filtered CSV.")

# Load accident data
acc = pd.read_csv(acc_file)
acc['Start_Time'] = pd.to_datetime(acc['Start_Time'])

# Keep only accidents during PEMS-BAY's time window (Jan–Jun 2017)
# Plus a small buffer on either side for nearest-neighbor matching
acc = acc[(acc['Start_Time'] >= '2016-12-31') &
          (acc['Start_Time'] <= '2017-07-01')].copy()
print(f"Accidents in PEMS-BAY window: {len(acc):,}")

# Weather observations: one record per accident with its timestamp + weather
weather_obs = acc[['Start_Time', 'Start_Lat', 'Start_Lng',
                   'Temperature(F)', 'Humidity(%)', 'Visibility(mi)',
                   'Wind_Speed(mph)', 'Weather_Condition']].copy()
weather_obs = weather_obs.rename(columns={'Start_Time': 'obs_time'})
weather_obs = weather_obs.sort_values('obs_time').reset_index(drop=True)
print(f"Weather observations: {len(weather_obs):,}")
```

## Cell 11 — Stage 1: Weather Enrichment
For each sensor reading (s, t), find the closest accident record in time within ±4 hours and ±50km.
Transfer its weather observations (temperature, humidity, visibility, wind speed, weather condition).

> ⏱️ **Runtime: 5–10 minutes**  
> Achieves **97.8% weather coverage** across all sensor readings.

```python
# For each sensor, do merge_asof against weather observations
# We do this per-sensor because merge_asof is a time-based match
print("Merging weather for each sensor (this takes ~5-10 min)...")

from tqdm import tqdm

long_df['timestamp'] = long_df['timestamp'].astype('datetime64[ns]')
weather_obs['obs_time'] = weather_obs['obs_time'].astype('datetime64[ns]')

results = []
unique_sensors = long_df['sensor_id'].unique()

for sid in tqdm(unique_sensors, desc='Weather merge'):
    sensor_rows = long_df[long_df['sensor_id'] == sid].sort_values('timestamp').copy()

    merged = pd.merge_asof(
        sensor_rows,
        weather_obs[['obs_time', 'Temperature(F)', 'Humidity(%)',
                     'Visibility(mi)', 'Wind_Speed(mph)', 'Weather_Condition']],
        left_on='timestamp', right_on='obs_time',
        direction='nearest',
        tolerance=pd.Timedelta('4h')
    )
    merged = merged.drop(columns=['obs_time'])
    results.append(merged)

merged_df = pd.concat(results, ignore_index=True)
print(f"\nAfter weather merge: {merged_df.shape}")
print(f"Weather coverage: {merged_df['Temperature(F)'].notna().mean()*100:.1f}%")
```

## Cell 12 — Stage 2: Accident Context Features
For each (sensor, timestamp), count accidents within 5km in the past 60 minutes.
Derives three features:
- `acc_count_60min` — number of nearby accidents in the lookback window
- `acc_max_severity` — maximum severity (1–4) of those accidents
- `acc_mins_since` — minutes since the most recent nearby accident

> ⏱️ **Runtime: 5–15 minutes**  
> Window computed on strictly past data — no look-ahead bias.

```python
print("Computing accident context features (this takes ~5-15 min)...")

from math import radians

NEARBY_KM = 5.0
WINDOW_MIN = 60

# Pre-compute radians for all accidents (once)
acc_lat_rad = np.radians(acc['Start_Lat'].values)
acc_lon_rad = np.radians(acc['Start_Lng'].values)
acc_times = acc['Start_Time'].values
acc_sev = acc['Severity'].values

# For each sensor, find accidents within 5 km (spatial filter, done once per sensor)
# Then for each timestep, find which of those happened in past 60 min

def compute_accident_features_for_sensor(sid):
    s_meta = meta_df[meta_df['sensor_id'] == sid].iloc[0]
    s_lat, s_lon = np.radians(s_meta['Latitude']), np.radians(s_meta['Longitude'])

    # Haversine distance to all accidents
    dlat = acc_lat_rad - s_lat
    dlon = acc_lon_rad - s_lon
    a = np.sin(dlat/2)**2 + np.cos(s_lat)*np.cos(acc_lat_rad)*np.sin(dlon/2)**2
    dist = 2 * 6371 * np.arcsin(np.sqrt(a))

    # Keep only nearby accidents (spatial filter)
    nearby_mask = dist <= NEARBY_KM
    nearby_times = acc_times[nearby_mask]
    nearby_sev = acc_sev[nearby_mask]

    # For each timestep of this sensor, count accidents in past 60 min
    sensor_rows = merged_df[merged_df['sensor_id'] == sid]
    timesteps = sensor_rows['timestamp'].values

    counts = np.zeros(len(timesteps), dtype=np.int32)
    max_sevs = np.zeros(len(timesteps), dtype=np.int32)
    mins_since = np.full(len(timesteps), WINDOW_MIN, dtype=np.float32)

    window_ns = np.timedelta64(WINDOW_MIN, 'm')

    for i, ts in enumerate(timesteps):
        window_start = ts - window_ns
        in_window = (nearby_times >= window_start) & (nearby_times <= ts)
        n = in_window.sum()
        counts[i] = n
        if n > 0:
            max_sevs[i] = nearby_sev[in_window].max()
            most_recent = nearby_times[in_window].max()
            mins_since[i] = (ts - most_recent) / np.timedelta64(1, 'm')

    return sensor_rows.index, counts, max_sevs, mins_since

# Initialize columns
merged_df['acc_count_60min'] = 0
merged_df['acc_max_severity'] = 0
merged_df['acc_mins_since'] = float(WINDOW_MIN)

for sid in tqdm(unique_sensors, desc='Accident features'):
    idx, counts, sevs, mins = compute_accident_features_for_sensor(sid)
    merged_df.loc[idx, 'acc_count_60min'] = counts
    merged_df.loc[idx, 'acc_max_severity'] = sevs
    merged_df.loc[idx, 'acc_mins_since'] = mins

print(f"\n✅ Accident features computed")
print(f"   Rows with nearby accident in past hour: {(merged_df['acc_count_60min'] > 0).sum():,}")
print(f"   Rows with no nearby accident:           {(merged_df['acc_count_60min'] == 0).sum():,}")
```

## Cell 13 — Add Time Features & Save
Derive temporal features (hour, day of week, weekend indicator).
Reorder columns into the final 16-feature schema and save as a compressed Parquet file.

```python
# Time-based features
merged_df['hour'] = merged_df['timestamp'].dt.hour
merged_df['dayofweek'] = merged_df['timestamp'].dt.dayofweek
merged_df['is_weekend'] = (merged_df['dayofweek'] >= 5).astype(int)

# Reorder columns into final 16-feature schema
col_order = [
    'timestamp', 'sensor_id', 'Latitude', 'Longitude', 'speed',
    'hour', 'dayofweek', 'is_weekend',
    'Temperature(F)', 'Humidity(%)', 'Visibility(mi)',
    'Wind_Speed(mph)', 'Weather_Condition',
    'acc_count_60min', 'acc_max_severity', 'acc_mins_since'
]
merged_df = merged_df[col_order]

# Save as parquet locally (compact, fast to reload)
output_path = os.path.join(dest_folder, 'merged_dataset_full.parquet')
merged_df.to_parquet(output_path, index=False, compression='snappy')

print(f"✅ Saved locally to: {output_path}")
print(f"   Rows: {len(merged_df):,}")
print(f"   Columns: {len(merged_df.columns)}")

# Show file size (Cross-platform Python alternative to `!ls -lh`)
file_size_mb = os.path.getsize(output_path) / (1024 * 1024)
print(f"   File size: {file_size_mb:.2f} MB")
```

Print shape, memory usage, feature coverage, and Pearson correlations with speed to verify the merged dataset is correct.
